Before Lamda

```python
def add(a, b):
    return a + b

print(add(3, 5)) 
```

After Lamda

```python
add_lambda = lambda a, b: a + b
print(add_lambda(3, 5))
```

```python
hello = lambda: "Namaste!"
square = lambda n: n * n
area = lambda length, width: length * width
greet = lambda name, msg="Hello": f"{msg}, {name}!"
total = lambda *nums: sum(nums)
total = lambda *nums: max(nums)


print(hello())             
print(square(6))            
print(area(5, 4))          
print(greet("Ram"))         
print(greet("Sita", "Hi"))  
print(total(10, 20, 30))
```

```python
check = lambda m: "Pass" if m >= 40 else "Fail"
print(check(55))    
print(check(30)) 
```

```python
check = lambda m=0: "Pass" if m >= 40 else "Fail"
print(check())
```

```python
square = lambda n: n * n
print(square(6))
```

```python
sign = lambda n: "yes" if n > 0 else "no"
print(sign(5)) 
```

```python
add = lambda a, b: a + b
shout = lambda word: word.upper() + "!"
is_even = lambda n: n % 2 == 0
bigger = lambda a, b: a if a > b else b

print(add(2, 3))
print(add("momo", "tea"))
print(shout("namaste"))
print(is_even(7))
print(bigger(10, 25))
print(add(5))
```

Key in Lamda

```python
words = ["banana", "kiwi", "apple", "fig" , "ab"]


print(sorted(words, key=len))

print(sorted(words, key=lambda w: w[-1]))
print(sorted(words, key=lambda w: w[1]))
```

```python
words = ["banana", "kiwi", "apple", "fig"]
last = lambda w: w[-1]

for w in words:
    print(w, "->", last(w))

print(sorted(words, key=last))
```

```python
students = [("Ram", 92), ("Sita", 78), ("Hari", 65)]

print(sorted(students))

print(sorted(students, key=lambda s: s[1]))

print(sorted(students, key=lambda s: s[1], reverse=True))
```

```python
students = [("Ram", 92), ("Sita", 78), ("Hari", 65)]

top = max(students, key=lambda s: s[1])
low = min(students, key=lambda s: s[1])

print(top)    
print(low)     

name, score = top
print(f"{name} to {score}")
```

```python
students = [
    {"name": "Ram", "marks": 92},
    {"name": "Sita", "marks": 78},
]
best = max(students, key=lambda s: s["marks"])
print(best)
print(best["name"]) 
```

```python
menu = {"momo": 150, "tea": 30, "lassi": 80}
cheap_first = sorted(menu.items(), key=lambda item: item[1], reverse=True)
print(cheap_first)
```

```python
words = ["Python", "is", "very", "fun"]
nums = [-7, 3, -1, 5]
team = [("Ram", 21), ("Sita", 19), ("Hari", 25)]

print(sorted(words, key=lambda w: len(w)))
print(sorted(nums, key=lambda n: abs(n)))
print(max(team, key=lambda p: p[1]))
print(min(team, key=lambda p: p[1])[0])
print(sorted(words, key=lambda w: w.lower()))
print(sorted(words))
```

map() and filter()

```python
prices = [100, 250, 80]


new_prices = []
for p in prices:
    new_prices.append(p + 20)
print(new_prices) 


new_prices = list(map(lambda p: p + 20, prices))
print(new_prices)       
```

```python
marks = "67 45 92".split()
print(marks)                   
print(list(map(int, marks))) 
```

```python
names = ["ram", "sita", "hari"]
print(list(map(lambda n: n.title(), names)))
```

```python
def square(n):
    return n * n
print(list(map(square, [1, 2, 3])))
```

```python
nums = [1, 2, 3]
result = map(lambda n: n * 10, nums) #map() works for one tim only

print(result)
print(list(result))


print(list(result))   # [10, 20, 30]
print(list(result))   # []   (already used up!)

# Tip: make the list once, use it many times
tens = list(map(lambda n: n * 10, nums))
print(tens, sum(tens)) 
```

filter(): keep only some items


```python
marks = [67, 35, 92, 28, 54]

# Day 08 way: loop + if
passed = []
for m in marks:
    if m >= 40:
        passed.append(m)
print(passed) 
```

```python
# filter + lambda way
passed = list(filter(lambda m: m >= 40, marks))
print(passed)
```

```python
words = ["momo", "tea", "chowmein", "dal"]
print(list(filter(lambda w: len(w) > 3, words)))
```

```python
marks = [67, 35, 92, 28, 54]

passed = filter(lambda m: m >= 40, marks)
bonus = map(lambda m: m + 5, passed)
final = list(bonus)

print(final)         
print(sum(final))     
print(max(final))     

print(list(map(lambda m: m + 5, filter(lambda m: m >= 40, marks))))
```

```python
nums = [1, 2, 3, 4, 5]
names = ["ram", "sita", "gopi"]

print(list(map(lambda n: n * n, nums)))
print(list(filter(lambda n: n > 2, nums)))
print(list(map(len, names)))
print(list(map(lambda n: n.upper(), names)))
print(list(filter(lambda n: "a" in n, names)))
print(list(map(lambda n: n + 1, nums)))
```
