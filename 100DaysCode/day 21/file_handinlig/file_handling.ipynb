File Handling

Writing to a file

```python
f = open("notes.txt", "w")      # open for writing

f.write("Hello Nepal\n")
f.write("Python is fun\n")

f.close() 
```

The better way: with open()


```python
with open("note.txt", "w") as f:
    f.write("Hello\n")
    f.write("Python\n")

# the file is closed here, by itself

print("Saved!")
```

writelines(): write a whole list

```python
fruits = ["apple\n", "mango\n", "banana\n"]

with open("fruits.txt", "w") as f:
    f.writelines(fruits)
```

```python
with open("fruit.txt", "w") as f:
    f.writelines(["apple", "mango", "banana"])
```

Reading a file: 3 ways

 read()

```python
with open("notes.txt", "r") as f:
    text = f.read()
print(text)
```

readlines():

```python
with open("notes.txt", "r") as f:
    lines = f.readlines()
print(lines) 
```

for loop: one line at a time

```python
with open("notes.txt", "r") as f:
    for line in f:
        print(line.strip())
```

readline(): one line per call

```python

with open("notes.txt", "r") as f:
    first = f.readline()       # 1st line
    second = f.readline()      # 2nd line
    third = f.readline()       # no more lines

print(first.strip())           
print(second.strip())        
print(third == "")  
```

Write vs append


```python
with open("notes.txt", "w") as f:
    f.write("New day\n")
```

```python
with open("notes.txt", "a") as f:
    f.write("Momo is tasty\n")
    f.write("Dal bhat power\n")
```

"r+": read and write together

```python

with open("diary.txt", "r+") as f:
    print(f.read())                   
    f.write("Monday: learned files\n") 
```

```python
with open("diary.txt", "r+") as f:
    f.write("SUN") 
```

When the file is not there

```python
with open("marks.txt", "r") as f:
    print(f.read())
```

```python
try:
    with open("marks.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("File not found. Check the name!")
```

Deleting a file

```python
import os

# os.remove("fruits.txt")          # fruits.txt is deleted

# Safe way: check first
if os.path.exists("fruit.txt"):
    os.remove("fruit.txt")
    print("Deleted!")
else:
    print("File not found")
```

Save marks, read them back

```python
students = {"Ram": 80, "Sita": 92, "Hari": 75}

# 1. Save: one student per line
with open("marks.txt", "w") as f:
    for name, marks in students.items():
        f.write(f"{name},{marks}\n")

# 2. Read back and find the total
total = 0
with open("marks.txt", "r") as f:
    for line in f:
        name, marks = line.strip().split(",")
        print(name, "got", marks)
        total = total + int(marks)

print("Total:", total)
```

Put It Together

Contact book

```python
class Contact:
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone

    def __str__(self):
        return f"{self.name}: {self.phone}"

contacts = [Contact("Ram", "9841000001"),
            Contact("Sita", "9851000002")]

# Save: str(c) uses __str__
with open("contacts.txt", "w") as f:
    for c in contacts:
        f.write(str(c) + "\n")

# Add one more later
with open("contacts.txt", "a") as f:
    f.write(str(Contact("Hari", "9861000003")) + "\n")

# Read and show
with open("contacts.txt", "r") as f:
    for line in f:
        print(line.strip())
```
