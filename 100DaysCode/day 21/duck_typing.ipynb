Duck Typing


Duck typing in code

```python
class Duck:
    def quack(self):
        print("Quack quack!")

class Person:
    def quack(self):
        print("I am saying quack!")

class Toy:
    def quack(self):
        print("Squeak quack!")

def make_it_quack(thing):
    thing.quack()             

make_it_quack(Duck())          
make_it_quack(Person())       
make_it_quack(Toy())  
```

When it can't quack

```python
class Stone:
    def roll(self):
        print("Stone rolls")

make_it_quack(Stone())
# AttributeError: 'Stone' object has no attribute 'quack'

# Safe version
def make_it_quack(thing):
    try:
        thing.quack()
    except AttributeError:
        print("This thing cannot quack")

make_it_quack(Duck())          # Quack quack!
make_it_quack(Stone())      
```

```python
class Phone:
    def charge(self):
        print("Phone is charging")

class Laptop:
    def charge(self):
        print("Laptop is charging")

class Book:
    def read(self):
        print("Reading the book")

def plug_in(device):
    try:
        device.charge()
    except AttributeError:
        print("Cannot charge this!")

for d in [Phone(), Laptop(), Book()]:
    plug_in(d)
```
