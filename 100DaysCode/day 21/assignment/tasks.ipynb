Task 01

```python
class Bird:
    def fly(self):
        print("Bird can fly")

class Plane:
    def fly(self):
        print("Plane can with passengers")

class Kite:
    def fly(self):
        print("Kite is a toy it aslo can fly")

def lift_off(thing):
    try:
        thing.fly()
    except AttributeError:
        print("This things cannot fly")

lift_off(Bird())
lift_off(Plane())
lift_off(Kite())
```

```python
class Stone:
    def roll(self):
        print("Stone can roll")

lift_off(Stone())
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} : {self.price}"

p = (Product("MO:MO", "Rs. 150"))
print(p)
```

```python
class ClassRoom:
    def __init__(self, names):
       self.names = names
    

    def __len__(self):
        return len(self.names)

room = ClassRoom(["Ram", "Hari", "Shyam", "Sita"])

print(len(room))
```

Task 04

```python
class Distance:
    def __init__(self, km):
        self.km = km

    def __add__(self, other):
        return self.km + other.km

    def __eq__(self, other):
       return self.km == other.km

d1 = Distance(12)
d2 = Distance(15)
d3 = Distance(15)


print(d1 + d2)
print(d1 == d2)
print(d2 + d3)
print(d2 == d3)
```
