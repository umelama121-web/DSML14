Magic Methods

```python
class Student:
    def __init__(self, name, marks):
        print("__init__ is running...")
        self.name = name
        self.marks = marks

s1 = Student("Ram", 80)
# __init__ is running...

print(s1.name)       # Ram
print(s1.marks)  
```

__str__: nice text for print()

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"{self.name} - {self.marks} marks"

s1 = Student("Ram", 80)
s2 = Student("Sita", 92)

print(s1)            # Ram - 80 marks
print(s2)            # Sita - 92 marks

text = str(s1)       # str() also uses __str__
print(text) 
```

__len__: make len() work

```python
class Basket:
    def __init__(self, fruits):
        self.fruits = fruits             # a list

    def __len__(self):
        return len(self.fruits)

b = Basket(["apple", "mango", "banana"])

print(len(b))    
```

__add__: make + work

```python
class Wallet:
    def __init__(self, money):
        self.money = money

    def __add__(self, other):
        return Wallet(self.money + other.money)

    def __str__(self):
        return f"Rs. {self.money}"

w1 = Wallet(500)
w2 = Wallet(300)

w3 = w1 + w2         # Python calls w1.__add__(w2)
print(w3) 
```

Compare objects: == and <

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks

    def __lt__(self, other):
        return self.marks < other.marks

ram = Student("Ram", 80)
sita = Student("Sita", 92)
hari = Student("Hari", 80)

print(ram == hari)   
print(ram == sita)   
print(ram < sita)    
print(sita < ram)   
```

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"Book: {self.title}"

    def __len__(self):
        return self.pages

    def __add__(self, other):
        return self.pages + other.pages

b1 = Book("Muna Madan", 120)
b2 = Book("Python Basics", 300)

print(b1)
print(len(b2))
print(b1 + b2)
```
