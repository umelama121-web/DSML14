Double kwargs

```python
def profile(**info):
    print(info)
    print(type(info))

profile(name='Alice', age=30, city='New York')
```

```python
def profile(**info):

    print(info.items())
    for key, value in info.items():
        print(f"{key}: {value}")
    print("========")

profile(name='Alice', age=30, city='New York')
profile(name='Bob', age=25, city='Los Angeles', profession='Engineer')
```

Use .get() for missing options

```python
def order_tea(**options):
    sugar = options.get("sugar", 1)
    milk = options.get("milk" "yes")
    print(f"Tea: sugar={sugar}, milk={milk}")
    print(f"Options sent: {len(options)}")

order_tea(sugar=2, milk="yes")
```

```python
def student(name, **marks):
    print(f"Student: {name}")
    print(marks)
    print(marks.items())
    for subject, mark in marks.items():
        print(f"  {subject}: {mark}")
    print(f"  Total: {sum(marks.values())}")

student("Gita", math=88, science=76, english=91)
```

```python
def names(**names):
    for key in names:
        print(f"{key}")
names(name="ram")
```

```python
def info(**details):
    return details

def count(**details):
    return len(details)

print(info(a=1, b=2))
print(count(x=10, y=20, z=30))
print(info())

def momo(plate, **extras):
    print(f"{plate} momo")
    for name, amount in extras.items():
        print(f"- {name} x{amount}")

momo("Buff", achar=2, soup=1)
momo("veg")
momo("Veg", "achar")
```

Using Both Together

```python
def demo(a, *args,  b, **kwargs):
    print("a      =", a)
    print("args   =", args)
    print("b      =",  b)
    print("kwargs =", kwargs)

demo(1, 2, 3, x=10,  b=3, y=20)
```

The order rule

```python
def order(customer, *items, delivery=False, **notes):
    print(customer, items, delivery, notes)

order("Ram", "momo", "tea", delivery=True, spicy="yes")
# Ram ('momo', 'tea') True {'spicy': 'yes'}
```

* opens a list


```python
def add(a, b, c):
    return a + b + c

nums = [10, 20, 30]

# Long way
print(add(nums[0], nums[1], nums[2]))   # 60

# Short way
print(add(*nums))  
```

print() + * = clean output

```python
fruits = ["apple", "mango", "banana"]

print(fruits)              # ['apple', 'mango', 'banana']
print(*fruits)             # apple mango banana
print(*fruits, sep=", ")   # apple, mango, banana

print(*fruits, sep="\n")
print(*fruits, sep="\t")
```


** opens a dictionary

```python
def profile(name, age, city):
    print(f"{name}, {age}, from {city}")

ram = {"name": "Ram", "age": 21, "city": "Pokhara"}

profile(**ram)
```

Unpack into *args

```python
def total(*prices):
    return sum(prices)

cart = [120, 80, 250]
more = (50, 30)

print(total(*cart))           
print(total(*cart, *more))    
print(total(*cart, 100))      
print(total(cart))
```
