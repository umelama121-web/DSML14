Abstract Class

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass                    # no work here

class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side * self.side

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

class Rectangle(Shape):

    def __init__(self, l, w):
        self.l = l
        self.w = w

    def area(self):
        return self.l * self.w

print(Square(4).area())     
print(Circle(2).area()) 
r1 = Rectangle(8, 9)
# print(Rectangle(4,2).area())
print(r1.l)
print(r1.w)
print(r1.area())
```

Every child has it, so loop safely


```python
shapes = [Square(4), Circle(2), Rectangle(6, 4)]

for s in shapes:
    print(s.area())
```

Normal methods are allowed too

```python
# from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name):
        self.name = name

    def eat(self):                       # normal
        print(f"{self.name} is eating")

    @abstractmethod
    def sound(self):                     # must write
        pass

class Cow(Animal):
    def sound(self):
        print(f"{self.name} says Moo!")

class Dog(Animal):
        def eat(self):
            print(f"{self.name} is eating")

        def sound(self):
             print(f"{self.name} says woof")



c1 = Cow("Gauri")
c1.eat()        # Gauri is eating
c1.sound()  
d1 = Dog("tommy")
d1.eat()
d1.sound()
```

```python
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class Esewa(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} with eSewa")

class Khalti(Payment):
    def pay(self, amount):
        print(f"Paid Rs. {amount} with Khalti")

class Cash(Payment):
    def pay(self, amount):
        print(f"paid Rs. {amount} with khalti")

Esewa().pay(500)
Khalti().pay(200)
Cash().pay(100)
```
