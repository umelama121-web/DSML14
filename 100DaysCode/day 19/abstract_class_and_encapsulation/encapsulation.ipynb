Encapsulation

Anyone can change anything

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

acc = BankAccount("Ram", 1000)
print(acc.balance)

acc.balance = -50000      
print(acc.balance)        

acc.balance = "hello"     
print(acc.balance) 
```

Public

```python
class Student:
    def __init__(self, name):
        self.name = name           # public

s1 = Student("Ram")
print(s1.name)        # Ram    (read from outside)

s1.name = "Hari"      # change from outside
print(s1.name) 
```

Protected: one underscore _

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance     # protected

acc = BankAccount("Ram", 1000)

print(acc._balance)    # 1000   (works, but you shouldn't)
acc._balance = 5   
print(acc._balance)
```

Private: two underscores __

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance    # private

    def show(self):
        print(f"{self.owner}: Rs. {self.__balance}")

acc = BankAccount("Ram", 1000)
acc.show()          # Ram: Rs. 1000   (inside: OK)

print(acc.__balance)
```

Getter and setter: the ATM buttons

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance

    def get_balance(self):                 # getter
        return self.__balance

    def deposit(self, amount):             # setter
        if amount > 0:
            self.__balance = self.__balance + amount
            print(f"Added Rs. {amount} new balance Rs. {self.__balance}")
        else:
            print("Amount must be more than 0")

acc = BankAccount("Ram", 1000)
acc.deposit(500)            # Added Rs. 500
acc.deposit(-200)           # Amount must be more than 0
print(acc.get_balance()) 
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.__marks = 0

    def get_marks(self):
        return self.__marks

    def set_marks(self, marks):
        if marks >= 0 and marks <= 100:
            self.__marks = marks
        else:
            print("Marks must be 0 to 100")

s1 = Student("Sita")
s1.set_marks(85)
s1.set_marks(500)          # Marks must be 0 to 100
print(s1.get_marks())
```

```python
class ATM:
    def __init__(self, pin):
        self.__pin = pin

    def check(self, pin):
        if pin == self.__pin:
            print("Welcome!")
        else:
            print("Wrong PIN")

    def change_pin(self, new_pin):
        self.__pin = new_pin
        print("pin updated sucessfully")



acc = ATM(1234)
acc.check(1111)
acc.check(1234)

acc.__pin = 1111
acc.check(1111)

acc.change_pin(1111)
acc.check(1111)
```

Put It Together

Bank accounts, both ideas

```python
from abc import ABC, abstractmethod

class Account(ABC):
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance          # private

    def add_interest(self):
        interest = self.__balance * self.rate() / 100
        self.__balance = self.__balance + interest

    def show(self):
        print(f"{self.owner}: Rs. {self.__balance}")

    @abstractmethod
    def rate(self):                       # must write
        pass

class Savings(Account):
    def rate(self):
        return 5

class Fixed(Account):
    def rate(self):
        return 10

s = Savings("Ram", 1000)
f = Fixed("Sita", 1000)
s.add_interest()
f.add_interest()
s.show()      # Ram: Rs. 1050.0
f.show()      # Sita: Rs. 1100.0
```
