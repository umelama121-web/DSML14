Give the object some data

```python
class Student:
    pass

s1 = Student()
s1.name = "Ram"     
s1.age = 20   
s1.home = "godawari"        

s2 = Student()
s2.name = "Sita"      
s2.age = 21   
s2.home = "pokhara"        

print(s1.name)        
print(s2.name)  
print(s1.age)
print(s2.age)  
print(s1.home)
print(s2.home)
```

Your first __init__

```python
class Student:
    def __init__(self, name, age):
        self.name = name 
        self.age = age

s1 = Student("Ram", 20)
s2 = Student("Sita", 23)
print(s1.name, s1.age)
print(s2.name, s2.age)
```

Starting values and changing data


```python
class Student:
    def __init__(self, name, age):
        self.name = name     
        self.age = age       
        self.marks = 0       

s1 = Student("Ram", 20)
print( s1.name, s1.age, s1.marks)    

s1.marks = 85     
print(s1.marks,s1.age, s1.marks) 
```

Each object has its own data

```python
s1 = Student("Ram", 20)
s2 = Student("Sita", 21)

s1.age = 25         # change only s1

print(s1.age)       # 25
print(s2.age) 
```

```python
class Mobile:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price
        self.battery = 100

m1 = Mobile("Samsung", 30000)
m2 = Mobile("Redmi", 18000)

m1.battery = 40

print(m1.brand, m1.battery)
print(m2.brand, m2.battery)
print(m1.price + m2.price)
```
