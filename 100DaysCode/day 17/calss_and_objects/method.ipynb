Methods

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hi, I am {self.name} I'm {self.age} years old.")

s1 = Student("Ram", 20)
s2 = Student("Sita", 21)

s1.greet() 
s2.name = "Gita"
s2.greet()  
```

Methods with inputs: a bank account

```python
class BankAccount:
    def __init__(self, owner):
        self.owner = owner
        self.balance = 0

    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"Added Rs. {amount} your new balance is {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Not enough money!")
        else:
            self.balance = self.balance - amount
            print(f"Took Rs. {amount} yor remaining balance is {self.balance}")

acc = BankAccount("Ram")
acc.deposit(1000)     
acc.withdraw(300)     
acc.withdraw(5000)     
print(acc.balance) 
```

A method that returns a value

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def is_square(self):
        return self.length == self.width

r1 = Rectangle(5, 3)
r2 = Rectangle(4, 4)

print(r1.area())         
print(r2.area())         
print(r1.is_square())    
print(r2.is_square()) 
```

```python
class Cart:
    def __init__(self, owner):
        self.owner = owner
        self.items = []

    def add(self, item):
        self.items.append(item)

    def count(self):
        return len(self.items)
    
    def remove(self, item):
        self.items.remove(item)
        print(f"{item} has been removed")

c1 = Cart("Ram")
c2 = Cart("Sita")

c1.add("momo")
c1.add("tea")
c2.add("chowmein")

print(c1.count())
print(c2.items)
c1.remove("tea")
print(f"{c1.owner} has {c1.count()} items")
```

Student report card

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.marks = []

    def add_mark(self, mark):
        self.marks.append(mark)
        print(f"{mark} is adde to {self.name} new marks list is {self.marks}")

    def average(self):
        return sum(self.marks) / len(self.marks)

    def report(self):
        avg = self.average()
        total = sum(self.marks)
        if avg >= 40:
            print(f"{self.name}: {avg} Pass total : {total}")
        else:
            print(f"{self.name}: {avg} Fail total : {total}")

s1 = Student("Ram")
s1.add_mark(80)
s1.add_mark(70)
s1.add_mark(0)
s1.add_mark(0)
s1.report() 
```
