Task 01

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.price = 0

    def info(self):
        print(f"{self.title} Book is written by {self.author} price is : ${self.price}.")



book1 = Book("Harry Potter and the Philosopher’s Stone", "J.K. Rowling")
book2 = Book("The Alchemist", "Paulo Coelho")
book3 = Book("Romeo and Juliet", "William Shakespeare")

book1.price = 500
book2.price = 450
book3.price = 300

book1.info()
book2.info()
book3.info()
```

Task 02

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

circle1 = Circle(5)
circle2 = Circle(7)
circle3 = Circle(9)

print(circle1.area())
print(circle2.area())
print(circle3.area())
```

Task 03

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increase(self):
        self.count += 1

    def reset(self):
        self.count = 0


c = Counter()

c.increase()
c.increase()
c.increase()

print(c.count)

c.reset()
print(c.count)
```

Task 04

```python
class Light:
    def __init__(self):
        self.is_on = False

    def switch(self):
        self.is_on = not self.is_on

light = Light()

light.switch()
print(light.is_on)

light.switch()
print(light.is_on)

light.switch()
print(light.is_on)

light.switch()
print(light.is_on)

```

Task 04

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def give_raise(self, amount):
        self.salary = self.salary + amount
        print(f"{self.name}'s salary has been raised by {amount} and now total salary is : {self.salary}")

staff1 = Employee("Ram", 50000)
staff2 = Employee("Sita", 60000)
staff3 = Employee("Hari", 45000)

staff1.give_raise(15000)
staff2.give_raise(10000)
staff3.give_raise(10000)

```
