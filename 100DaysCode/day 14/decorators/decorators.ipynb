Decorators, Recursion and Generators

```python
def my_decorator(func):
    def wrapper():
        print("----- start -----")
        func()
        print("----- end -----")
    return wrapper

def say_hello():
    print("Hello!")

say_hello = my_decorator(say_hello)
say_hello()

def say_hi():
    print("Hi!")

say_hi = my_decorator(say_hi)
say_hi()
```

```python
@my_decorator
def say_bye():
    print("Bye!")

say_bye()
```

```python
def my_decorator(func):
    print("1. decorating", func.__name__)
    def wrapper():
        print("3. before")
        func()
        print("5. after")
    return wrapper

@my_decorator
def say_hello():
    print("4. Hello!")

print("2. now we call it")
say_hello()
```

Decorate any function

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Calling", func.__name__)
        result = func(*args, **kwargs)
        print("Done")
        return result
    return wrapper

@my_decorator
def add(a, b):
    return a + b

total = add(3, 4)
print(total)
```

```python
logged_in = False


def login_required(func):
    def wrapper(*args, **kwargs):
        if not logged_in:
            print("Please log in first!")
            return
        return func(*args, **kwargs)
    return wrapper

@login_required
def see_balance():
    print("Balance: Rs. 5000")

see_balance()      
logged_in = True
see_balance()  
```

Stack decorators


```python
def stars(func):
    def wrapper():
        print("*****")
        func()
        print("*****")
    return wrapper

def dashes(func):
    def wrapper():
        print("-----")
        func()
        print("-----")
    return wrapper

@stars
@dashes
def hello():
    print("Hello!")

hello()
```

```python

def double_it(func):
    def wrapper(n):
        return func(n) * 2
    return wrapper

def say(func):
    def wrapper():
        print("Before")
        func()
    return wrapper

@double_it
def add_ten(n):
    return n + 10

@say
def hi():
    print("Hi")

print(add_ten(5))
hi()
print(add_ten(0))
add_ten()
```
