__init__, self and methods


```python
class Student:
    school = "Skill Shikshya"     # class attribute: shared

    def __init__(self, name, marks):
        self.name = name          # object attribute: own copy
        self.marks = marks

    def greet(self):
        print("Hi, I am", self.name)

    def result(self):
        return "Pass" if self.marks >= 40 else "Fail"

s1 = Student("Ram", 85)       # __init__ runs here
s1.greet()                    # Hi, I am Ram
print(s1.result())            # Pass
print(s1.school)      
```

Inheritance

overriding, super()

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def sound(self):
        print("Some sound")
    def eat(self):
        print(self.name, "is eating")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)    # parent sets name
        self.breed = breed        # child adds breed
    def sound(self):              # override
        print("Woof!")

d = Dog("Tommy", "Husky")
d.eat()      
d.sound()  
print(d.breed)

```

Abstraction & Encapsulation

Abstract class


```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    def show(self):                 # normal method
        print("Area:", self.area())

class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self):                 # must write this
        return self.side * self.side

class Rectangle(Shape):
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth
    def area(self):                 # must write this
        return self.length * self.breadth

    
Square(4).show()    
Rectangle(4, 6).show()

# Shape() 
```

Public, protected, private

Getter and setter

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks       # private

    def get_marks(self):           # getter
        return self.__marks

    def set_marks(self, marks):    # setter
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Marks must be 0 to 100")

s = Student("Ram", 80)
s.set_marks(150) 
s.set_marks(90)       # Marks must be 0 to 100
print(s.get_marks())    # 80
print(s.__marks)        # Private value cant access without getter method
```

Polymorphism

One call, many answers

```python
class Staff:
    def pay(self):
        return 0

class Teacher(Staff):
    def pay(self):
        return 40000

class Driver(Staff):
    def pay(self):
        return 25000

total = 0
for person in [Teacher(), Driver(), Teacher()]:
    total += person.pay()       # same call, different answer
print(total)    
```

Method overloading

```python
class Calculator:
    def add(self, a, b=0, c=0):     # way 1: default values
        return a + b + c

    def total(self, *nums):         # way 2: *args
        return sum(nums)

c = Calculator()
print(c.add(5))               
print(c.add(5, 3))            
print(c.add(5, 3, 2))         
print(c.total(1, 2, 3, 4))    
```

Duck typing

```python
class Duck:
    def quack(self):
        print("Quack quack!")

class Person:
    def quack(self):
        print("I am saying quack!")

class Toy:
    def quack(self):
        print("Squeak quack!")

def make_it_quack(thing):
    thing.quack()                 # no type check

for t in [Duck(), Person(), Toy()]:
    make_it_quack(t)
```

Magic Methods

Teach your class print, len, + and ==

```python
class Wallet:
    def __init__(self, money):
        self.money = money
    def __str__(self):
        return f"Wallet(Rs {self.money})"
    def __len__(self):
        return self.money
    def __add__(self, other):
        return Wallet(self.money + other.money)
    def __eq__(self, other):
        return self.money == other.money
    def __lt__(self, other):
        return self.money < other.money

w1, w2 = Wallet(500), Wallet(300)
print(w1 + w2)     
print(len(w1))     
print(w1 == w2)    
print(w2 < w1)    
```

Put It Together

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
    def get_balance(self):
        return self.__balance
    def __str__(self):
        return f"{self.owner}: Rs {self.__balance}"

class Savings(Account):
    def deposit(self, amount):
        super().deposit(amount + 10)    # bonus

for acc in [Account("Ram", 100), Savings("Sita")]:
    acc.deposit(50)
    print(acc)

a = Account("Hari", 150)
a.__balance = 0
print(a.get_balance())
```
