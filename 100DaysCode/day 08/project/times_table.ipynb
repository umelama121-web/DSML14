```python

num = int(input("Which table? "))
up_to = int(input("Up to which number? "))

for i in range(1, up_to + 1):
    print(f"{num} x {i} = {num * i}")
# total = sum(num * i for i in range(1, up_to + 1))
# print(f"Total: {total}")
total = 0
for i in range(1, up_to + 1):
    result = num * i
    total += result
print(f"Total: {total}")
    



# Your job:
# 1. ask "Up to?" and use it in range()
# 2. add up all the answers and print the total
```
