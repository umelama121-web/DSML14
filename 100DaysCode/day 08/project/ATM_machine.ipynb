```python
balance = 1000
pin = "1234"
tries = 0

while  True:
    print("Welcome to the ATM")
    user_pin = input("Enter your pin: ").strip()
    if user_pin != pin:
        tries += 1
        print(f"Incorrect PIN. You have {3 - tries} tries left.")
        if tries == 3:
            print("Too many incorrect attempts. Exiting.")
            break
        continue
    print("1. Balance  2. Deposit  3. Withdraw  4. Exit")
    choice = input("Choose (1-4): ").strip()
    if choice == "1":
        print(f"Balance: Rs. {balance}")
    elif choice == "2":
        amount = int(input("Amount to deposit: "))
        if amount <= 0:
            print("Invalid deposit amount.")
        else:
            balance += amount
            print(f"New balance: Rs. {balance}")
            break
    elif choice == "3":
        amount = int(input("Amount to withdraw: "))
        if amount > balance:
            print("Not enough money")
        else:
            balance -= amount
            print(f"Take your cash. Left: Rs. {balance}")
            print("Thank you! Bye")
            break
    elif choice == "4":
        print("Thank you! Bye")
        break
else:
    print("Thank you , visit again")

# Your job:
# 1. ask for a PIN first, only 3 tries (like Project 02)
# 2. don't allow a deposit of 0 or less
```
