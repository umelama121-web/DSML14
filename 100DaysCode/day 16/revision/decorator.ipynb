Decorator

```python
def run_twice(func):     
    func()                 
    func()                 

def hello():
    print("Hello!")

def add():
    print(2 + 2)

run_twice(hello)  
run_twice(add)      
```

```python
def make_box():
    def inner():
        print("I am inner")
    return inner          

new_func = make_box()     
new_func() 
```

Long way

```python
def add_line(func):           # 1. take a function
    def wrapper():              # 2. make a cover for it
        print("----------")     #    extra work before
        func()                  #    run the real function
        print("----------")     #    extra work after
    return wrapper              # 3. give the cover back

def say_hi():
    print("Hi!")

say_hi = add_line(say_hi)       # put the cover on
say_hi()
```

Short and right way to use Decorator

```python
@add_line
def say_hi():
    print("Hi!")

say_hi()
```

```python
def my_decorator(func):
    def wrapper():
        # extra work BEFORE
        print("Before the function runs")
        func()
        print("After the function runs")
        # extra work AFTER
    return wrapper          # no () here!

@my_decorator
def my_function():
    print("my real work")

my_function()
```

```python
def show(func):
    def wrapper(*args):
        print("Start")
        answer = func(*args)     # pass the inputs on
        print("End")
        return answer            # give the answer back
    return wrapper

@show
def add(a, b):
    return a + b

print(add(2, 3))
```

Example: password check

```python
def need_password(func):
    def wrapper():
        password = input("Password: ")
        if password == "1234":
            func()                  # correct: run it
        else:
            print("Wrong password!")
    return wrapper

@need_password
def show_balance():
    print("Balance: Rs. 5000")

show_balance()
```

```python
def shout(func):
    def wrapper():
        print("HEY!")
        func()
        print("BYE!")
    return wrapper

@shout
def eat():
    print("Eating momo")

@shout
def sleep():
    print("Sleeping")

eat()
sleep()
```
