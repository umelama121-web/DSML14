Generators

```python
def three_numbers():
    yield 1
    yield 2
    yield 3

g = three_numbers()
# print(g)        # <generator object ...>  not the numbers!

# print(next(g))  
# print(next(g))
# print(next(g))


for n in g:
    print(n)
```

```python
def cook():
    print("Cooking plate 1...")
    yield "Plate 1"
    print("Cooking plate 2...")
    yield "Plate 2"
    print("Cooking plate 3...")

g = cook()            # nothing prints yet

print(next(g))
print(next(g))
```

yield inside a loop

```python
def count_up(limit):
    n = 1
    while n <= limit:
        yield n          # give n, pause here
        n = n + 1        # continue from here next time

for x in count_up(3):
    print(x)



print(list(count_up(5)))  
```

A generator works only once

```python
def count_up(limit):
    n = 1
    while n <= limit:
        yield n
        n = n + 1

g = count_up(3)
print(list(g))      # [1, 2, 3]
print(list(g))      # []   empty! already used

g = count_up(3)     # make a new one
print(list(g))
```

List vs generator, same answer

```python
def squares_list(n):
    result = []
    for i in range(1, n + 1):
        result.append(i * i)
    return result

# Generator: makes ONE at a time
def squares_gen(n):
    for i in range(1, n + 1):
        yield i * i

print(squares_list(4))         # [1, 4, 9, 16]
print(list(squares_gen(4)))
```
