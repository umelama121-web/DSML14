Recursion

```python
def countdown(n):
    if n == 0:              # Rule 1: base case
        print("Go!")
        return              # stop here
    print(n)
    countdown(n - 1)        # Rule 2: smaller step

countdown(3)
```

```python
def sum_to(n):
    if n == 1:                  # base case
        return 1
    return n + sum_to(n - 1)    # smaller step

print(sum_to(4))     # 10   (4 + 3 + 2 + 1)
print(sum_to(5))  
```

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(4))
```

Loop and recursion

```python
def sum_loop(n):
    total = 0
    for i in range(1, n + 1):
        total = total + i
    return total

# With recursion
def sum_to(n):
    if n == 1:
        return 1
    return n + sum_to(n - 1)

print(sum_loop(4))    # 10
print(sum_to(4)) 
```

```python

def stars(n):
    if n == 0:
        return
    print("*" * n)
    stars(n - 1)

def double_sum(n):
    if n == 0:
        return 0
    return 2 + double_sum(n - 1)

stars(3)
print(double_sum(4))
```
