```python
def get_grade(mark):
    if mark >= 80:
        return "A"
    elif mark >= 60:
        return "B"
    elif mark >= 40:
        return "C"
    else:
        return "F"

def average(marks):
    return sum(marks.values()) / len(marks)

def topper(marks):
    return max(marks, key=marks.get)

marks = {"Ram": 78, "Sita": 92, "Hari": 35, "Gita": 64}

for name, mark in marks.items():
    print(f"{name}: {mark} -> {get_grade(mark)}")

print(f"Class average: {average(marks)}")

# count = sum(1 for mark in marks.values() if mark >= 40)
count = 0
for mark in marks.values():
    if mark >= 40:
        count += 1

print(f"NUmber of the student who got 40 or more marks: {count} Students.")
topper_student = topper(marks)
print(f"The topper is: {topper_student} with marks: {marks[topper_student]}")




# Your job:
# 1. write count_passed(marks): how many got 40 or more?
# 2. write topper(marks): return the name with the top mark
```
