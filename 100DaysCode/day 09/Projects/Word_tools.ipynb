Word tools

```python
def count_vowels(text):
    count = 0
    for letter in text.lower():
        if letter in "aeiou":
            count += 1
    return count

def is_palindrome(word):
    word = word.lower()
    return word == word[::-1]


def count_space(text):
    count = 0
    for letter in text:
        if letter == " ":
            count += 1
    return count

def first_letter(text):
    words = text.split()
    result = ""

    for word in words:
        result += word[0]
    return result

print(count_vowels("I love Nepal"))   # 5
print(is_palindrome("Madam"))         # True
print(is_palindrome("Nepal")) 

count = count_space("My name is umesh tamang") 
print(count)
word = first_letter("I'm from godawari")
print(word)

# Your job:
# 1. write count_spaces(text)
# 2. write first_letters(text): "I love Nepal" gives "ILN"
#    (hint: .split() from Day 04)
```
