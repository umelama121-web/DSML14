Temperature Converter

```python

from string.templatelib import convert


def c_to_f(c):
    return (c * 9 / 5) + 32

def f_to_c(f):
    return (f - 32) * 5 / 9

def is_hot(c):
    return c > 30

print(c_to_f(100))     # 212.0
print(f_to_c(50))      # 10.0


# temp = float(input("Temperature in Celsius: "))
# print(f"That is {c_to_f(temp)} F")

choice = input("what do you you want to convert \n 1. C to F 2. F to C ").strip().lower()
temp = float(input("Temperature you want to convert: "))
if choice == "1":
    print(f"That is {c_to_f(temp): .2f} F")
elif choice == "2":
    print(f"That is {f_to_c(temp): .2f} C")
else:
    print("We dont have that option")


if is_hot(temp):
    print("It is so hot")
else:
    print("It's not hot")
    
    


# Your job:
# 1. ask: C to F, or F to C? Then call the right one (Day 07)
# 2. write is_hot(c) that returns True above 30
```
