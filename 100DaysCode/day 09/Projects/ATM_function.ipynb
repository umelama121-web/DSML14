```python
def show_menu():
    print("1. Balance  2. Deposit  3. Withdraw  4. Exit")

def deposit(balance, amount):
    if amount <= 0:
        print("Amount must be more than 0")
        return balance
    return balance + amount

def withdraw(balance, amount):
    if amount > balance:
        print("Not enough money")
        return balance
    return balance - amount

def check_pin(pin):
    return pin == 1234






balance = 1000
tries = 0

while tries <3:
    pin = int(input("Enter your pin: "))
    if check_pin(pin):
        print("correct pin")
        break
    else:
        print("Wrong pin. Try again")
        tries += 1
        print(f"You have {3 - tries} attempts left")

if tries == 3:
    print("Yor are out of attempts")
else:
    while True:
        print("Welcome to ATM")
        
        show_menu()
        choice = input("Choose (1-4): ").strip()

        if choice == "1":
            print(f"Balance: Rs. {balance}")
        elif choice == "2":
            amount = int(input("Amount to deposit: "))
            balance = deposit(balance, amount)
            print(f"Your new balance is : {balance}")
        elif choice == "3":
            amount = int(input("Amount to withdraw: "))
            balance = withdraw(balance, amount)
            print(f"Youe new balance is : {balance}")
            print("Thank you. Please visit again")
            break
        elif choice == "4":
            print("Thank you! Bye")
            break
        else:
            print("Please choose 1 to 4")

# Your job:
# 1. write check_pin(pin) that returns True or False
# 2. ask for the PIN first, only 3 tries (Day 08)
```
