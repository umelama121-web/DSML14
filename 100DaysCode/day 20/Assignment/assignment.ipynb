Task 01

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name):
        self.name = name
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print(f"{self.name} says : Woof")

class Cat(Animal):
    def sound(self):
        print(f"{self.name} says : Meow")

class Duck(Animal):
    def sound(self):
        print(f"{self.name} says : Quack")

for s in [Dog("Dante"), Cat("Lucy"), Duck("Daffy")]:
    s.sound()
```

```python
class Guitar:
    def play_music(self):
        print("Guitar is playing a music")

class Madal:
    def play_music(self):
        print("Madal is giving a beat")

class Flute:
    def play_music(self):
        print("Flute is adding a sweet music")

def play(instrument):
    instrument.play_music()

g = Guitar()
m = Madal()
f = Flute()

play(g)
play(m)
play(f)
```

Task 03

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Square(Shape):
    def __init__(self, s):
        self.s = s

    def area(self):
        return self.s * self.s

class Rectangle(Shape):
    def __init__(self, l, w):
        self.l = l
        self.w = w

    def area(self):
        return self.l * self.w

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

shapes = [Square(4), Rectangle(2, 3), Circle(5)]

total = 0

for s in shapes:
    total += s.area()

print("Total area :", total )

```

Task 05

```python
from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send(self,message):
        pass

class Sms(Notification):
    def send(self,message):
        print("sending mesage through Sms :", message)

class Email(Notification):
    def send(self, message):
        print("sending email :", message)

class Viber(Notification):
    def send(self, message):
        print("sending message through Viber:", message)


s = Sms()
e = Email()
v = Viber()

s.send("Hello")
e.send("Check your email")
v.send("How are you")
```
