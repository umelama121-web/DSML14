Method Overloading

Method overloading: quick recap

```python
class Calculator:
    def add(self, a, b, c=0):          # default value
        return a + b + c

    def total(self, *args):            # any number
        return sum(args)

calc = Calculator()
print(calc.add(2, 3))                
print(calc.add(2, 3, 4))            

print(calc.total(1, 2))              
print(calc.total(1, 2, 3, 4, 5))     
```

```python
class Greet:
    def hello(self, name="Friend", times=1):
        for i in range(times):
            print(f"Hello {name}")

g = Greet()
g.hello()
g.hello("Ram")
g.hello("Sita", 2)
```

Momo shop bill

```python
from abc import ABC, abstractmethod

class Food(ABC):
    def __init__(self, plates):
        self.plates = plates

    @abstractmethod
    def price(self):                 
        pass

class Momo(Food):
    def price(self):
        return self.plates * 150

class Chowmein(Food):
    def price(self):
        return self.plates * 120

class Pizza(Food):
    def price(self):
        return self.plates * 500

order = [Momo(2), Chowmein(1), Pizza(1)]

total = 0
for f in order:
    total = total + f.price()      

print(f"Total bill: Rs. {total}")    
```
