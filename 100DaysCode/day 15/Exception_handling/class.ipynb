Exceptions and Exception Handling

first try and except

```python
try:
    age = int(input("Enter your age: "))
    print("Next year you will be", age + 1)
except ValueError:
    print("Please type a number!")

print("Program continues...")

```

```python
try:
    a = int(input("First number: "))
    b = int(input("Second number: "))
    print("Answer:", a / b)
except ValueError:
    print("Please type numbers only.")
except ZeroDivisionError:
    print("You can't divide by zero.")
```

```python
try:
    num = int("12abc")
except ValueError as e:
    print("Error:", e)
# Error: invalid literal for int() with base 10: '12abc'

# Two errors, one block: use a tuple
try:
    print(10 / 0)
except (ValueError, ZeroDivisionError) as e:
    print("Problem:", e)
# Problem: division by zero
```

```python
def check_age(text):
    try:
        age = int(text)
    except ValueError:
        print("Not a number!")
    else:
        print("Your age is", age)
    finally:
        print("--- check done ---")

check_age("25")
# Your age is 25
# --- check done ---

check_age("abc")
# Not a number!
# --- check done ---
```

```python
nums = [10, 20, 30]

try:
    print(nums[1])
    print(nums[5])
    print("A")
except IndexError:
    print("B")
else:
    print("C")
finally:
    print("D")

try:
    print(int("7") + 3)
except ValueError:
    print("E")
```

Use It with What We Know

```python
while True:
    try:
        age = int(input("Enter your age: "))
        break
    except ValueError:
        print("That's not a number. Try again.")

print("Thanks! Your age is", age)
```

Write it once, use it often


```python
def get_number(message):
    while True:
        try:
            return int(input(message))
        except ValueError:
            print("Numbers only, please!")

def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Can't divide by zero"

x = get_number("First number: ")
y = get_number("Second number: ")
print(safe_divide(x, y))
```

How raise works

```python
age = -5

if age < 0:
    raise ValueError("Age can't be negative")

print("Age saved") 
```

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age can't be negative")
    print("Age saved:", age)

set_age(20)


try:
    set_age(-5)
except ValueError as e:
    print("Error:", e)
```

```python
try:
    age = int(input("Enter your age: "))

    if age < 18 or age >= 0:
        raise ValueError("Age must positive and be 18 or above!")

    print("Access granted!")

except ValueError as e:
    print("Error:", e)

print("Program continues...")
```
