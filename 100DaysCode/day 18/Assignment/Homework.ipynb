Task 03

```python
class Shape:
    def area(self):
        return 0

class Square(Shape):
    def __init__(self, s):
        self.s = s

    def area(self):
        return self.s * self.s

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r

s = Square(7)
c = Circle(8)
shape = Shape()


print(" Shape area:", shape.area())
print("Square area:", s.area())
print("Circle area:", c.area())
```

Task 03

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age



class Employee(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.salary = salary

    def info(self):
        print(f"Name:{self.name}\nAge:{self.age}\nSalary:{self.salary}")


person = Employee("Ram", 28, 50000)
person.info()


```

Task 04

```python
class Vehicle():
    def __init__(self, brand):
        self.brand = brand

    def info(self):
        print(f"{self.brand}")


class Car(Vehicle):
    def __init__(self, brand, wheel):
        super().__init__(brand)
        self.wheel = wheel

    def info(self):
        print(f"{self.brand} has {self.wheel} wheels.")

class ElectricCar(Car):
    def __init__(self, brand, wheel, range):
        super().__init__(brand, wheel)
        self.range = range

    def info(self):
        print(f"{self.brand} has {self.wheel} wheel and gives {self.range} km range.")

car1 = Vehicle("Toyta")
car2 = Car("Range Rover", 4)
car3 = ElectricCar("BYD", 4, 500)

car1.info()
car2.info()
car3.info()


```
