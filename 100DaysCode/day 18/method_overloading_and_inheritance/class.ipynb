Python keeps only the last one

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def add(self, a, b, c):      # replaces the first add
        return a + b + c

calc = Calculator()
print(calc.add(1, 2, 3))   # 6

print(calc.add(1, 2))
```

Way 1: default values

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

calc = Calculator()

print(calc.add(2, 3))       # 5    (c = 0)
print(calc.add(2, 3, 4)) 
```

Different inputs, different work

```python
class Student:
    def __init__(self, name):
        self.name = name

    def greet(self, time=""):
        if time == "":
            print(f"Hi, I am {self.name}")
        else:
            print(f"Good {time}, I am {self.name}")

s1 = Student("Ram")
s1.greet()              
s1.greet("morning") 
```

Way 2: *args for any number

```python
class Calculator:
    def add(self, *numbers):
        return sum(numbers)

calc = Calculator()

print(calc.add(5))               
print(calc.add(5, 10))           
print(calc.add(5, 10, 15, 20)) 
```

Back to the momo shop

```python
class MomoShop:
    def order(self, item="buff momo", plates=1):
        price = 150 * plates
        print(f"{plates} plate {item}: Rs. {price}")

shop = MomoShop()

shop.order()                   
shop.order("chicken momo")      
shop.order("veg momo", 3) 
```

```python
class Greeter:
    def hello(self, name="friend", times=1):
        for i in range(times):
            print(f"Hello, {name}!")

g = Greeter()

g.hello()
g.hello("Sita")
g.hello("Hari", 2)
```
