Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating")

class Dog(Animal):
    pass                  

d1 = Dog("Tommy")
d2 = Dog("goldy")
d2.eat()
d1.eat()            
print(d1.name)
```

A child adds its own methods

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} is eating")

class Dog(Animal):
    def bark(self):
        print(f"{self.name} says Woof!")

class Cat(Animal):
    def meow(self):
        print(f"{self.name} says Meow!")

d1 = Dog("Tommy")
c1 = Cat("Kitty")

d1.eat()      
d1.bark()      
c1.eat()       
c1.meow()  
```

The "is a" test

```python
d1 = Dog("Tommy")
a1 = Animal("Lion")

print(isinstance(d1, Dog))      
print(isinstance(d1, Animal))   
print(isinstance(a1, Dog)) 
```

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
        self.wheels = 4

    def start(self):
        print(f"{self.brand} started")

class Bike(Vehicle):
    def ring(self):
        print("Tring tring!")

b1 = Bike("Honda")
b1.start()
b1.ring()
b1.wheels = 2
print(b1.wheels)
```

Types of Inheritance

Single inheritance

```python
class Father:
    def house(self):
        print("I have a house")

class Son(Father):
    def bike(self):
        print("I have a bike")

s = Son()
s.house()     
s.bike() 
```

Multilevel inheritance

```python
class Grandfather:
    def land(self):
        print("Land from grandfather")

class Father(Grandfather):
    def house(self):
        print("House from father")

class Son(Father):
    def bike(self):
        print("My own bike")

s = Son()
s.land()    
s.house()     
s.bike() 
f = Father()
f.house()
f.land()
```

Hierarchical inheritance

```python
class Person:
    def __init__(self, name):
        self.name = name

    def walk(self):
        print(f"{self.name} is walking")

class Student(Person):
    def study(self):
        print(f"{self.name} is studying")

class Teacher(Person):
    def teach(self):
        print(f"{self.name} is teaching")

s = Student("Ram")
t = Teacher("Sita")

s.walk()     
t.walk()     
s.study()    
t.teach()   
```

Multiple inheritance

```python
class Father:
    def height(self):
        print("Tall like father")

class Mother:
    def eyes(self):
        print("Brown eyes like mother")

class Child(Father, Mother):
    pass

c = Child()
c.height()   
c.eyes() 
```

Hybrid inheritance


```python
class Person:
    def breathe(self):
        print("Breathing")

class Student(Person):
    def study(self):
        print("Studying")

class Worker(Person):
    def work(self):
        print("Working")

class Intern(Student, Worker):
    pass

i = Intern()
i.breathe()   
i.study()    
i.work()   
```

Put It Together

Savings account

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount
        print(f"Added Rs. {amount}")

    def show(self):
        print(f"{self.owner}: Rs. {self.balance}")

class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, rate=5):
        super().__init__(owner, balance)
        self.rate = rate

    def add_interest(self):
        interest = self.balance * self.rate / 100
        self.deposit(interest)

acc = SavingsAccount("Ram", 1000)
acc.deposit(500)       
acc.add_interest()     
acc.show()            
```
