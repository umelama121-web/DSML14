Overriding and super()
first override


```python
class Animal:
    def __init__(self, name):
        self.name = name

    def sound(self):
        print("Some sound")

class Dog(Animal):
    def sound(self):             
        print("Woof!")

class Cat(Animal):
    def sound(self):             
        print("Meow!")

class Fish(Animal):
    pass                         
Dog("Tommy").sound()    
Cat("Kitty").sound()     
Fish("Nemo").sound()  
```

One loop, many animals

```python
animals = [Dog("Tommy"), Cat("Kitty"), Fish("Nemo")]

for a in animals:
    print(a.name, end=": ")
    a.sound()
```

The child needs extra data

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, school):
        self.school = school       # only school is saved

s1 = Student("Ram", 20, "Skill Shikshya")
print(s1.school)    # Skill Shikshya

print(s1.name)
```

super() to the rescue

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, school):
        super().__init__(name, age)   # parent saves name, age
        self.school = school          # child saves school

s1 = Student("Ram", 20, "Skill Shikshya")

print(s1.name)      
print(s1.age)     
print(s1.school) 
```

Keep the parent's work, add more

```python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"Hi, I am {self.name}")

class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def introduce(self):
        super().introduce()               
        print(f"I teach {self.subject}")   

t1 = Teacher("Shikhar", "Python")
t1.introduce()
```

Fix the bike

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand
        self.wheels = 4

    def info(self):
        print(f"{self.brand} has {self.wheels} wheels")

class Bike(Vehicle):
    def __init__(self, brand):
        super().__init__(brand)
        self.wheels = 2

car = Vehicle("Toyota")
bike = Bike("Honda")

car.info()
bike.info()
print(bike.brand)
```
