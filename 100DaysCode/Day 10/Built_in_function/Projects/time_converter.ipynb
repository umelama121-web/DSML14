```python
def convert(seconds):
    minutes, secs = divmod(seconds, 60)
    hours, minutes = divmod(minutes, 60)
    return hours, minutes, secs

def to_second(h, m, s):
    return h * 3600 + m * 60 + s


h, m, s = convert(3725)
print(f"{h} h {m} m {s} s")      

h = input("enter hours :")
m = input("enter minutes :")
s = input("enter second :")



total = to_second(int(h), int(m), int(s))

print(f"total second is : {total}")
      
# Your job:
# 1. ask the user for seconds with int(input())
# 2. write to_seconds(h, m, s) that goes the other way
```
