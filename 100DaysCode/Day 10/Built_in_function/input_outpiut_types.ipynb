```python
marks = [10, 20, 30, 40, 50]
total = 0
biggest = marks[0]
for m in marks:
    total += 1
    if m > biggest:
        biggest = m
print(total, biggest)
```

enumerate

```python
fruits = ["apple", "mango", "banana"]

# Old way (Day 08 homework)
count = 1
for fruit in fruits:
    print(f"{count}. {fruit}")
    count += 1

# New way
for count, fruit in enumerate(fruits, start=-5):
    print(f"{count}. {fruit}")
```

Zip

```python
names = ["Ram", "Sita", "Hari"]
marks = [78, 92, 35]

for name, mark in zip(names, marks):
    print(f"{name} got {mark}")

student_marks = dict(zip(names, marks))
print(student_marks)
```

all and any

```python
list1 =[True, False, True, False]
print(all(list1))
print(any(list1))
```

```python
marks = [78, 92, 35, 64]
passed = []
for m in marks:
    passed.append(m >= 40)
print(passed)
print(all(passed))
print(any(passed))
```
